# Описание проекта

В проекте проведён анализ поведения пользователей образовательной платформы
и выполнена сегментация пользователей с помощью алгоритмов кластеризации.

## Используемые технологии

- Python
- Pandas
- NumPy
- Scikit-learn
- Matplotlib
- Seaborn
- CatBoost

## Кластеризация

Для сегментации пользователей использовались алгоритмы:

- KMeans
- DBSCAN

Количество кластеров для KMeans определялось с помощью метода локтя
и метода силуэта.

В результате было выделено 4 основных сегмента пользователей.

## Результаты

### Кластер 0 — самые активные пользователи

Это самые активные пользователи платформы. Они часто заходят на платформу
и используют практически все её разделы.

### Кластер 1 — пользователи, работающие с вебинарами

Основная активность этих пользователей связана с проведением вебинаров.

### Кластер 2 — наименее активные пользователи

Пользователи редко заходят на платформу и в основном просматривают информацию.

### Кластер 3 — пользователи, которые чаще работают с материалами

Основная активность этих пользователей связана с поиском, просмотром
и использованием учебных материалов.
